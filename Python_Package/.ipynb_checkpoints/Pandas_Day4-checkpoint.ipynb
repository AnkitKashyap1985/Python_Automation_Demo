```python
import pandas as pd
```

# Merge (SQL Joins)

```python
df_emp = pd.read_csv("data/employees.csv")
```

```python
df_emp.head()
```

```python
df_dept = pd.read_csv("data/department.csv")
```

```python
df_dept.head()
```

```python
df_sal = pd.read_csv("data/salary.csv")
```

```python
df_sal.head()
```

# Inner Join

```python
# SQL : Select * from employees as emp Inner join department as dept on emp.deptno=dept.deptid
ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='inner')
```

```python
ans_df
```

# Left Join

```python
# SQL : Select * from employees as emp left join department as dept on emp.deptno=dept.deptid
ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='left')
```

```python
ans_df
```

# Right Join

```python
# SQL : Select * from employees as emp Right join department as dept on emp.deptno=dept.deptid
ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='right')
```

```python
ans_df
```

```python
# same can be achived using left join by changing the dataframe sequence
# SQL : Select * from department as dept left join employees as emp on emp.deptno=dept.deptid

ans_df = df_dept.merge(df_emp,left_on='deptid',right_on='deptno',how='left')
```

```python
ans_df
```

# Full Outer Join

```python
# SQL : Select * from employees as emp full join  department as dept on emp.deptno=dept.deptid

ans_df = df_emp.merge(df_dept,left_on='deptno',right_on='deptid',how='outer')
```

```python
ans_df
```

# Concat ()

```python
# concat() used to concatenate either along rows(axis=0) or along columns (axis=1)
# along axis = 0
#df1(5 rows)
#df2(4 rows)

# along axis = 1 (along columns)
#df1(5 rows) df2(4 rows)
```

```python
pd.concat([df_emp,df_dept], axis=0)
```

```python
pd.concat([df_dept,df_emp],axis=1)
```

```python
# Concate 3 dataframes
pd.concat([df_emp,df_dept,df_sal], axis=1)
```

```python
# Concate 3 dataframes
pd.concat([df_emp,df_dept,df_sal], axis=0)
```

# Union and Union All

```python
# Union all
# SQL: 
# Select eno, ename, deptno from emp
# Union all
# Select eno, ename, deptno from dept

pd.concat([df_emp,df_emp1])
```

```python
# Union
# SQL: 
# Select eno, ename, deptno from emp
# Union
# Select eno, ename, deptno from dept

pd.concat([df_emp,df_emp1]).drop_duplicates()
```

# Intersection

```python
# Intersection
# SQL: 
# Select eno, ename, deptno from emp
# Intersect
# Select eno, ename, deptno from dept

df_emp.merge(df_emp1,how='inner')
```
