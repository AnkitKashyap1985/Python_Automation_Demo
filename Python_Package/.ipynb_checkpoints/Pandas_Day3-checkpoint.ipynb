```python
import pandas as pd
```

```python
df = pd.read_csv("emp.csv")
df.head()
```

# Group by

```python
# Aggregate functions (Sum, min, max, count, avg)
```

```python
#Q1. Count the employees in each department.
# expected O/P : deptno, count_employees
# SQL : select deptno, count(*) from emp(df) group by deptno

aggr_data = df.groupby('deptno')['eno'].count()
aggr_data
```

```python
type(aggr_data)
```

```python
#Convert the series in to dataframe
aggr_data = aggr_data.reset_index()
aggr_data
```

```python
type(aggr_data)
```

```python
#Rename eno column into count_employees
aggr_data.rename(columns={"eno":"count_employees"},inplace=True)
```

```python
aggr_data
```

```python
#Q2. Get me Total count of employees, Max salary, Min salary and Total salary in each department.
# SQL:  select count(eno) as employees_count, max(salary) as max_salary, min(salary) as min_salary, sum(salary) as total salary
# group by deptno

agg_data = df.groupby ('deptno')['salary'].agg(['count', 'max', 'min', 'sum'])
```

```python
type(agg_data)
```

```python
agg_data
```

```python
#Flatten the data to add indexes
agg_data = agg_data.reset_index()
```

```python
agg_data.rename(columns={'count':"Employee_Count",'max':'Max_Salary','min':"Min_Salary",'sum':"Total_Salary"},inplace=True)
```

```python
agg_data
```

```python
# Write this summary data into emp_summary.csv
agg_data.to_csv("emp_summary.csv",index=False)
```

```python
# Updated data sets (emp.csv)
df = pd.read_csv('emp.csv')
```

```python
df.head()
```

```python
#Q3. Get me employees count in each location and within each department
# expected O/P:
# location  deptno  emp_count
# Bangalore  10        2
# Bangalore  20        5
# Chennai    20        5
# Chennai    30        8
# SQL Equivalent: select location, deptno, count(eno) as emp_count from emp group by location, deptno

data = df.groupby(['location', 'deptno'])['eno'].count().reset_index()
data.rename(columns = {'eno':"emp_count"},inplace=True)
data
```

```python
# Q4. Get me total salary in each deparment and except for location = "Mumbai" and total salary must be more than 55K
# SQL: select deptno,sum(salary) as total_sal from emp where location != "Mumbai" group by deptno having sum(salary)>50K

filter_location = df['location'] != "Mumbai"
df_filtered = df[filter_location]
df_agg = df_filtered.groupby('deptno')['salary'].sum()
df_agg1 = df_agg.reset_index()
```

```python
having_filter = df_agg1['salary'] >50000
df_agg1[having_filter]
```

```python
# way2 (lamba function)
lambda_fil = df_agg.apply(lambda x:x>50000)
df_agg[lambda_fil].reset_index()
```
