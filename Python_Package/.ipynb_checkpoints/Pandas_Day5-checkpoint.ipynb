```python
# Q1. Assume my source : Oracle db and target : mysql db, identify which product_id's are not loaded in target but available in source
```

```python
# Create a Oracle and Mysql connection string
mysql_tgt_conn = create_engine("mysql+pymysql://root:Admin%40123@localhost:3306/pandas_practice")
oracle_src_conn = create_engine("oracle+oracledb://system:admin@localhost:1521/xe")
```

```python
df_target = pd.read_sql("select * from product",mysql_tgt_conn)
```

```python
df_target.head()
```

```python
df_source = pd.read_sql("select * from product",oracle_src_conn)
```

```python
df_source.head()
```

```python
# Check if source is having any extra product_id
# SQL: Select product_id from source
#      minus
#      select product_id from target
fil = df_source['product_id'].isin(df_target['product_id'])
df_source[~fil]
```

```python
# Check if target is having any extra product_id than source
# SQL: Select product_id from target
#      minus
#      select product_id from source
fil = df_target['product_id'].isin(df_source['product_id'])
df_source[~fil]
```
